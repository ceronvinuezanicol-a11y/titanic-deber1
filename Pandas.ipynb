```python
import pandas as pd

# 1. Cargamos el archivo oficial del Titanic
df = pd.read_csv("titanic.csv")

# 2. Calculamos los datos pedidos
total_pasajeros = len(df)
passengers_unicos = df["PassengerId"].nunique()

# 3. Mostramos los resultados en pantalla
print(f"Total de pasajeros en el dataset: {total_pasajeros}")
print(f"Total de PassengerId distintos: {passengers_unicos}")

if total_pasajeros == passengers_unicos:
  print(
      "Conclusión: Sí coinciden exactamente, lo que indica que cada fila es"
      " un pasajero único."
  )
else:
  print("Conclusión: No coinciden.")
```

```python
import pandas as pd

# Cargar el dataset 
df = pd.read_csv("titanic.csv")

print("=== PREGUNTA 2: Pasajeros por clase ===")
conteo_clases = df["Pclass"].value_counts().sort_index()
print(conteo_clases)
print(f"La clase con más pasajeros registrados es la Tercera Clase (Pclass 3).\n")

print("=== PREGUNTA 3: Pasajeros por puerto de embarque ===")
# dropna=False para ver también si hay valores vacíos (NaN)
puertos = df["Embarked"].value_counts(dropna=False)
print(puertos)
print(
    "Nota sobre valores faltantes: Hay 2 pasajeros con el puerto de embarque"
    " sin registrar (NaN).\n"
)

print("=== PREGUNTA 4: Tarifas (Fare) por clase ===")
tarifas_clase = df.groupby("Pclass")["Fare"].agg(["mean", "min", "max"])
print(tarifas_clase)
```

Pregunta 3: Se identificaron 2 registros sin puerto asignado. Estos valores no se eliminaron ni se inventaron; se cuantificaron explícitamente utilizando el parámetro dropna=False para dejar constancia de su existencia dentro del total de pasajeros.

```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv("titanic.csv")

print("=== PREGUNTA 5: Pasajeros de cada clase por puerto de embarque ===")
# Usamos crosstab para hacer una tabla cruzada
p5 = pd.crosstab(df["Pclass"], df["Embarked"], dropna=False)
print(p5)
print("\n")

print("=== PREGUNTA 6: Supervivientes (1) y fallecidos (0) por clase ===")
p6 = pd.crosstab(df["Pclass"], df["Survived"])
# Renombramos columnas para mayor claridad
p6.columns = ["Fallecidos (0)", "Sobrevivientes (1)"]
print(p6)
print("\n")

print(
    "=== PREGUNTA 7: Pasajeros según cantidad de familiares a bordo ==="
)
# Sumamos SibSp (hermanos/cónyuges) + Parch (padres/hijos)
df["TotalFamiliares"] = df["SibSp"] + df["Parch"]


def clasificar_familiares(n):
  if n == 0:
    return "0 familiares"
  elif n == 1:
    return "1 familiar"
  else:
    return "2 o más familiares"


df["CategoriaFamilia"] = df["TotalFamiliares"].apply(clasificar_familiares)
print(
    df["CategoriaFamilia"].value_counts()[
        ["0 familiares", "1 familiar", "2 o más familiares"]
    ]
)
print("\n")

print(
    "=== PREGUNTA 8: Edades (menores de 18, 18 o más, y sin registrar) por"
    " clase ==="
)


import pandas as pd

df = pd.read_csv("titanic.csv")


# Clasificar edades
def clasificar_edad(edad):
  if pd.isna(edad):
    return "Edad Sin Registrar"
  elif edad < 18:
    return "Menores de 18 años"
  else:
    return "Adultos (18 años o más)"


df["CategoriaEdad"] = df["Age"].apply(clasificar_edad)

# Crear la tabla cruzada
p8 = pd.crosstab(df["Pclass"], df["CategoriaEdad"])[
    ["Menores de 18 años", "Adultos (18 años o más)", "Edad Sin Registrar"]
]

# Cambiar nombres de las filas para que no sean solo números
p8.index = ["Primera Clase", "Segunda Clase", "Tercera Clase"]
p8.index.name = "Clase del Pasajero"

print(p8)

```

Pregunta 8: Los valores faltantes en la edad (especialmente notorios en tercera clase con 136 casos) no se rellenaron con promedios artificiales para evitar sesgar el análisis. En su lugar, se agruparon en una categoría independiente denominada "Edad Sin Registrar" en la tabla cruzada

```python
import pandas as pd

# Cargar el dataset
df = pd.read_csv("titanic.csv")

print("=== PREGUNTA 9: Análisis de códigos de boleto (Ticket) ===")
total_tickets_distintos = df["Ticket"].nunique()
conteo_tickets = df["Ticket"].value_counts()
tickets_multiples = conteo_tickets[conteo_tickets > 1]
ticket_mas_frecuente = conteo_tickets.idxmax()
frecuencia_maxima = conteo_tickets.max()

print(f"Total de códigos de boleto distintos: {total_tickets_distintos}")
print(
    f"Boletos asociados a más de un pasajero: {len(tickets_multiples)}"
)
print(
    f"El boleto que aparece en más registros es '{ticket_mas_frecuente}'"
    f" (repetido {frecuencia_maxima} veces)."
)
print("\n")

print("=== PREGUNTA 10: Registros con y sin cabina por clase ===")
# Verificamos qué registros tienen la columna Cabin nula (sin cabina)
df["Sin_Cabina"] = df["Cabin"].isna()

# Agrupamos por Pclass calculando cuántos tienen cabina, cuántos no, y el total
resumen_cabinas = df.groupby("Pclass").agg(
    Con_Cabina=("Cabin", lambda x: x.notna().sum()),
    Sin_Cabina=("Cabin", lambda x: x.isna().sum()),
    Total_Pasajeros=("Pclass", "size"),
)
print(resumen_cabinas)
```

Pregunta 10: Debido a que una gran cantidad de pasajeros carecían de este dato, los valores nulos se evaluaron mediante métodos de detección de nulos (.isna()) para separarlos y contrastarlos de forma exacta frente a los registros que sí contaban con un código de cabina por clase.
